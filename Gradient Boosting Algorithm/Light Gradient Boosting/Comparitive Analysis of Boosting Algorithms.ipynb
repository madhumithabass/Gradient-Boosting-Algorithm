```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.ensemble import AdaBoostClassifier,GradientBoostingClassifier
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score
```

## Data Collection and Understanding

```python
credit_data = pd.read_csv("credit_card_clean.csv")
credit_data
```

```python
credit_data.shape
```

```python
credit_data.isnull().sum()
```

```python
credit_data.dtypes
```

## Data Preparation

```python
### Convert the columns sex, education, marriage into numbers by preprocessing (encoding technique) for the ML model to understand

le = LabelEncoder()

credit_data["SEX"] =  le.fit_transform(credit_data["SEX"])
credit_data["EDUCATION"] =  le.fit_transform(credit_data["EDUCATION"])
credit_data["MARRIAGE"] =  le.fit_transform(credit_data["MARRIAGE"])
```

```python
credit_data
```

```python
## Split X and y

## DEFAULT is the output column here in this dataset

X = credit_data.drop("DEFAULT", axis = 1)
y = credit_data["DEFAULT"]
```

```python
X
```

```python
y
```

```python
## Split the dataset into training and testing data

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,
    random_state=123,
    shuffle=True,
    stratify=y,)
```

```python
print(X_train.shape,y_train.shape)
print(X_test.shape,y_test.shape)
```

## Model Building

```python
adaboost =  AdaBoostClassifier()
gboost= GradientBoostingClassifier(n_estimators= 50,max_depth=3)
xgboost = XGBClassifier(n_estimators=50,max_depth=3)
lightgboost = LGBMClassifier(n_estimators=50,max_depth=3, force_col_wise= True)
```

```python
%%time

adaboost_classifier =  adaboost.fit(X_train,y_train)
```

```python
%%time

gb_classifier = gboost.fit(X_train,y_train)
```

```python
%%time

xgb_classifier = xgboost.fit(X_train,y_train)
```

```python
%%time

lightgb_classifier = lightgboost.fit(X_train,y_train)
```

## Model Training and Model Evaluation

```python
y_pred_ada = adaboost_classifier.predict(X_train)
y_pred_gb = gb_classifier.predict(X_train)
y_pred_xgb = xgb_classifier.predict(X_train)
y_pred_lightgb = lightgb_classifier.predict(X_train)
```

### AdaBoost Classifier

```python
print("Confusion Matrix:", confusion_matrix(y_train,y_pred_ada))
print("Accuracy Score:", round(accuracy_score(y_train,y_pred_ada),4))
print("Precision Score:", round(precision_score(y_train,y_pred_ada),4))
print("Recall Score:", round(recall_score(y_train,y_pred_ada),4))
```

### Gradient Boosting Classifier

```python
print("Confusion Matrix:", confusion_matrix(y_train,y_pred_gb))
print("Accuracy Score:", round(accuracy_score(y_train,y_pred_gb),4))
print("Precision Score:", round(precision_score(y_train,y_pred_gb),4))
print("Recall Score:", round(recall_score(y_train,y_pred_gb),4))
```

## Xtreme Gradient Boosting Classifier

```python
print("Confusion Matrix:", confusion_matrix(y_train,y_pred_xgb))
print("Accuracy Score:", round(accuracy_score(y_train,y_pred_xgb),4))
print("Precision Score:", round(precision_score(y_train,y_pred_xgb),4))
print("Recall Score:", round(recall_score(y_train,y_pred_xgb),4))
```

```python
print("Confusion Matrix:", confusion_matrix(y_train,y_pred_lightgb))
print("Accuracy Score:", round(accuracy_score(y_train,y_pred_lightgb),4))
print("Precision Score:", round(precision_score(y_train,y_pred_lightgb),4))
print("Recall Score:", round(recall_score(y_train,y_pred_lightgb),4))
```

## Model Testing and Model Evaluation

```python
y_test_pred_ada = adaboost_classifier.predict(X_test)
y_test_pred_gb = gb_classifier.predict(X_test)
y_test_pred_xgb = xgb_classifier.predict(X_test)
y_test_pred_lightgb = lightgb_classifier.predict(X_test)
```

```python
print("Confusion Matrix:", confusion_matrix(y_test,y_test_pred_xgb))
print("Accuracy Score:", round(accuracy_score(y_test,y_test_pred_xgb),4))
print("Precision Score:", round(precision_score(y_test,y_test_pred_xgb),4))
print("Recall Score:", round(recall_score(y_test,y_test_pred_xgb),4))
```
