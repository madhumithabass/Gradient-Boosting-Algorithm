## Import Necessary Libraries

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, accuracy_score,precision_score,recall_score
```

## Data Collection and understanding

```python
breast_data = load_breast_cancer()
breast_data
```

```python
breast_data_df = pd.DataFrame(breast_data["data"], columns= breast_data["feature_names"])
breast_data_df["Target"] = breast_data.target
```

```python
breast_data_df
```

## Data Preparation

```python
breast_data_df.shape
```

```python
breast_data_df.isnull().sum()
```

## Data Preprocessing

```python
## Split Input X and out y


X = breast_data_df.drop("Target", axis = 1)
y = breast_data_df["Target"]
```

```python
X
```

```python
y
```

```python
scaled  = StandardScaler()

scaled_X =  scaled.fit_transform(X)
print(scaled_X)
```

```python
X_train,X_test,y_train,y_test = train_test_split(scaled_X,y,test_size=0.2,random_state=123,shuffle=True,stratify=y)
```

```python
print(X_train.shape,y_train.shape)
print(X_test.shape,y_test.shape)
```

## Model Building

```python
GB_classifier = GradientBoostingClassifier(learning_rate=0.03,n_estimators=30,max_depth=3)

GB_classifier.fit(X_train,y_train)
```

## Model Training and Model Evaluation

```python
y_pred_train = GB_classifier.predict(X_train)

print(confusion_matrix(y_pred_train,y_train))
```

```python
print(accuracy_score(y_pred_train,y_train))
```

```python
print(precision_score(y_pred_train,y_train))
```

```python
print(recall_score(y_pred_train,y_train))
```

## Model Testing and Model Evaluation

```python
y_pred_test = GB_classifier.predict(X_test)

print(confusion_matrix(y_pred_test,y_test))
```

```python
print(accuracy_score(y_pred_test,y_test))
```

```python
print(precision_score(y_pred_test,y_test))
```

```python
print(recall_score(y_pred_test,y_test))
```
